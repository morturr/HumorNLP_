```python
import pandas as pd

names_datasets = ['amazon', 'dadjokes', 'headlines', 'one_liners', 'yelp_reviews']
path = '../Data/new_humor_datasets/balanced/{DATASET}/data.csv'
```

### Calculate mean & std of text length

```python
for dataset_name in names_datasets:
    df = pd.read_csv(path.format(DATASET=dataset_name))
    df['text_length'] = df['text'].apply(lambda s: len(s))
    mean = df['text_length'].mean()
    std = df['text_length'].std()
    print(f'Statistics of text length for dataset {dataset_name}')
    print(f'{"%.2f" % mean} +- {"%.2f" % std}')
```
